```python
import os
import json
from pathlib import Path

# Create directory structure locally in Colab runtime
os.makedirs("data/raw", exist_ok=True)
os.makedirs("data/processed", exist_ok=True)
os.makedirs("processors", exist_ok=True)
os.makedirs("src", exist_ok=True)
os.makedirs("results/tables", exist_ok=True)
os.makedirs("results/figures", exist_ok=True)

# Create processor_A.json
proc_a_data = {
    "name": "processor_A",
    "num_qubits": 5,
    "coupling_map": [[0, 1], [1, 2], [2, 3], [3, 4]],
    "basis_gates": ["rz", "sx", "x", "cx"],
    "ports": [0, 4]
}
with open("processors/processor_A.json", "w") as f:
    json.dump(proc_a_data, f, indent=4)

# Create processor_B.json
proc_b_data = {
    "name": "processor_B",
    "num_qubits": 5,
    "coupling_map": [[0, 1], [1, 2], [2, 3], [3, 4], [4, 0]],
    "basis_gates": ["rz", "sx", "x", "cx"],
    "ports": [0, 2]
}
with open("processors/processor_B.json", "w") as f:
    json.dump(proc_b_data, f, indent=4)

# Create empty __init__.py
Path("src/__init__.py").touch()

print("Folders and processor JSON files created successfully!")
```

```python
# 1. src/processors.py
with open("src/processors.py", "w") as f:
    f.write('''import json
from pathlib import Path

def load_processor(path):
    with Path(path).open("r", encoding="utf-8") as f:
        return json.load(f)
''')

# 2. src/metrics.py
with open("src/metrics.py", "w") as f:
    f.write('''def circuit_metrics(qc):
    ops = qc.count_ops()
    two_q_names = {"cx", "cz", "ecr", "swap", "rxx", "ryy", "rzz"}
    return {
        "depth": qc.depth(),
        "num_qubits": qc.num_qubits,
        "total_ops": sum(ops.values()),
        "two_qubit_ops_estimate": sum(n for g, n in ops.items() if g.lower() in two_q_names),
        "swap_count": ops.get("swap", 0),
        "operations": dict(ops)
    }
''')

# 3. src/routing.py
with open("src/routing.py", "w") as f:
    f.write('''from qiskit.compiler import transpile
from qiskit.providers.fake_provider import GenericBackendV2

def get_transpiled_circuit(logical_qc, processor_config):
    coupling_map = processor_config.get("coupling_map")
    basis_gates = processor_config.get("basis_gates")
    num_qubits = processor_config.get("num_qubits")

    backend = GenericBackendV2(
        num_qubits=num_qubits,
        coupling_map=coupling_map,
        basis_gates=basis_gates
    )
    return transpile(logical_qc, backend=backend, optimization_level=3)
''')

# 4. src/problem.py
with open("src/problem.py", "w") as f:
    f.write('''from qiskit import QuantumCircuit
import numpy as np

def get_classifier_circuit(feature_vector, weights):
    num_qubits = len(feature_vector)
    qc = QuantumCircuit(num_qubits, num_qubits)
    for i, val in enumerate(feature_vector):
        qc.rx(val * np.pi, i)
    for i in range(num_qubits - 1):
        qc.cx(i, i + 1)
    for i, w in enumerate(weights[:num_qubits]):
        qc.rz(w, i)
    qc.measure(range(num_qubits), range(num_qubits))
    return qc
''')

# 5. src/visualization.py
with open("src/visualization.py", "w") as f:
    f.write('''import matplotlib.pyplot as plt
import networkx as nx
from pathlib import Path

def plot_processor_graph(processor_config, save_path=None):
    G = nx.Graph()
    G.add_nodes_from(range(processor_config["num_qubits"]))
    G.add_edges_from(processor_config["coupling_map"])
    plt.figure(figsize=(6, 4))
    nx.draw(G, with_labels=True, node_size=800, node_color="lightblue", font_weight="bold")
    plt.title(f"Hardware Topology: {processor_config['name']}")
    if save_path:
        Path(save_path).parent.mkdir(parents=True, exist_ok=True)
        plt.savefig(save_path)
    plt.close()

def plot_comparison_chart(metrics_a, metrics_b, save_path=None):
    labels = ["Depth", "Total Ops", "2-Qubit Ops", "SWAPs"]
    vals_a = [metrics_a["depth"], metrics_a["total_ops"], metrics_a["two_qubit_ops_estimate"], metrics_a["swap_count"]]
    vals_b = [metrics_b["depth"], metrics_b["total_ops"], metrics_b["two_qubit_ops_estimate"], metrics_b["swap_count"]]
    x = range(len(labels))
    width = 0.35
    plt.figure(figsize=(8, 5))
    plt.bar([p - width/2 for p in x], vals_a, width, label="Processor A")
    plt.bar([p + width/2 for p in x], vals_b, width, label="Processor B")
    plt.xlabel("Circuit Metrics")
    plt.ylabel("Count")
    plt.title("A/B Architecture Benchmark Comparison")
    plt.xticks(x, labels)
    plt.legend()
    if save_path:
        Path(save_path).parent.mkdir(parents=True, exist_ok=True)
        plt.savefig(save_path)
    plt.close()
''')

print("All source files written successfully inside Colab!")
```

```python
import sys
# Ensure current directory is in python path
if '.' not in sys.path:
    sys.path.append('.')

from src.processors import load_processor
from src.problem import get_classifier_circuit
from src.routing import get_transpiled_circuit
from src.metrics import circuit_metrics
from src.visualization import plot_processor_graph, plot_comparison_chart

# Load hardware configurations
proc_a = load_processor("processors/processor_A.json")
proc_b = load_processor("processors/processor_B.json")

# Build logical circuit for PS-Q1
features = [0.5, 0.2, 0.8, 0.1, 0.9]
weights = [0.1, 0.3, 0.2, 0.4, 0.5]
logical_circuit = get_classifier_circuit(features, weights)

# Transpile across architectures
compiled_a = get_transpiled_circuit(logical_circuit, proc_a)
compiled_b = get_transpiled_circuit(logical_circuit, proc_b)

# Extract metrics
metrics_a = circuit_metrics(compiled_a)
metrics_b = circuit_metrics(compiled_b)

print("--- PROCESSOR A METRICS ---")
print(metrics_a)
print("\n--- PROCESSOR B METRICS ---")
print(metrics_b)

# Generate and save comparison charts
plot_processor_graph(proc_a, "results/figures/processor_A_topo.png")
plot_processor_graph(proc_b, "results/figures/processor_B_topo.png")
plot_comparison_chart(metrics_a, metrics_b, "results/figures/ab_comparison.png")
print("\nBenchmark completed successfully! Figures saved to results/figures/")
```
